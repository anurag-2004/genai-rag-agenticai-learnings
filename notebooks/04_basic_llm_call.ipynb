```python
from openai import OpenAI
model= "gpt-4.1"
```

```python
llm= OpenAI()
```

```python
response= llm.chat.completions.create(
    model=model,
    messages=[
        {"role": "user", "content": "What is LLM. Explain me in 200 words"}
    ],
    # max_tokens=60
)
```

```python
response.choices[0].message.content
```

```python
response= llm.chat.completions.create(
    model=model,
    messages=[
        {"role": "user", "content": "What is LLM. Explain me in 200 words"}
    ],
    stream=True
)
```

```python
response
```

```python
from time import sleep
for chunk in response:
    print(chunk.choices[0].delta.content, end="")
    sleep(0.03)
```
