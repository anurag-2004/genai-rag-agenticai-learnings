```python
import os
import gc
from langchain_community.document_loaders import PyMuPDFLoader
```

```python
PATH= "../data/pdf"
```

```python
def iter_files(path:str):
    allItems = os.scandir(path)
    for item in allItems:
        if item.is_file() and item.name.lower().endswith(".pdf"):
            yield item.path


def stream_chunks(filePath:str):
    loader= PyMuPDFLoader(file_path=filePath)
    docs= loader.lazy_load()
    for page in docs:
        yield page.metadata

def ingest():
    batch=[]
    for path in iter_files(PATH):
        for chunk in stream_chunks(path):
            batch.append(chunk)

            if len(batch)>=200:
                ### generate the embeddings and store in vectore store
                batch.clear()
                gc.collect()

    if batch:
        batch.clear()
        gc.collect()
        pass
```

```python
ingest()
```

```python
import pymupdf
import pytesseract
from PIL import Image
import io
```

```python
def get_image_text(page)->str:
    images= page.get_pixmap()
    with open("scanned_invoice.png","wb") as img:
       img.write(images.tobytes())
    image= Image.open("./scanned_invoice.png")
    text= pytesseract.image_to_string(image)
    return text
```

```python
docs= pymupdf.open("../data/pdf/medical_report.pdf")
```

```python
for page in docs:
    page_text= page.get_text()
    image_text= get_image_text(page)

    final_text= page_text + "\n" + image_text
```
