```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.tools import tool
from langchain.agents import create_agent
from langchain_groq import ChatGroq
```

```python
@tool
def get_current_weather(city:str) -> str:
    """
    Get the current weather in a given city.
    Args:
        city (str): The name of the city.
    """
    # For the purpose of this example, we'll return a hardcoded response.
    # In a real implementation, you would call a weather API here.
    return f"The current weather in {city} is sunny with a temperature of 25°C."

@tool
def google_search(query:str) -> str:
    """
    Perform a Google search and return the top result.
    Args:
        query (str): The search query.
    """
    # For the purpose of this example, we'll return a hardcoded response.
    # In a real implementation, you would call a search API here.
    return f"Address is Bhopal"
```

```python
llm = ChatGroq(model= "llama-3.3-70b-versatile")
```

```python
agent= create_agent(model=llm, tools=[get_current_weather, google_search])
```

```python
res= agent.invoke(
    {"messages" : [{"role": "user", "content": "What is the current weather in New York?"}]}
)
```

```python
for msg in res["messages"]:
    msg.pretty_print()
```
