```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm= ChatOpenAI(model="gpt-4.1")
```

```python
response=llm.invoke([{"role": "user", "content": "Hello"}])
```

```python
response.content
```

```python
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
```

```python
parser.invoke(response)
```

```python
def toUpper(text: str) -> str:
    return text.upper()
```

```python
first_chain= llm | parser | toUpper
```

```python
res= first_chain.invoke([{"role": "user", "content": "Hello"}])
```

```python
res
```

```python
from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
from langchain_openai import ChatOpenAI
```

```python
from pydantic import BaseModel,Field
```

```python
class ModelResponse(BaseModel):
    name: str = Field(description="The name of the person")
    age: int = Field(description="The age of the person")
    email: str = Field(description="The email of the person")
    city: str = Field(description="The city of the person")
```

```python
llm= ChatOpenAI(model="gpt-4.1")
llm_with_str_out= llm.with_structured_output(ModelResponse)
```

```python
query= "I am Anurag Mohanty from Bhubaneswar, My age is 21 and email is mohantyanurag2004@gmail.com. Could you please give me age,name,email and city"
```

```python
res= llm_with_str_out.invoke([{"role": "user", "content": query}])
```
